section-b logical & conceptual

```python

marks ={'anu':85,
        'rahul':72,
        'akhil':91,
        'meera':65,
        'neha':78
    }
```

```python
#1)
highest_mark=max(marks,key=marks.get)
print('highest mark:', highest_mark,'-',marks[highest_mark])
```

```python
#2)
lowest_mark=min(marks,key=marks.get)
print('lowest_mark:',lowest_mark,'-',marks[lowest_mark])
```

```python
#3)
avarage=sum(marks.values()) / len(marks)
print('avarage mark:',avarage)
```

```python
#4)
passed_students= {student:mark for student,mark in marks.items() if mark>=75}
print('students scoring 75 or above:',passed_students)
```
